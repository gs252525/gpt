```python
from datasets import load_dataset
TRAIN_DATASET = "codeparrot/codeparrot-clean"
SEED = 1557

def get_batch(batch_size, dataset): 
    batch_code = []
    for row in dataset: 
        code = row.get('content') 
        batch_code.append(code) 
        if len(batch_code) == batch_size: 
            yield batch_code 
            batch_code = []

dataset = load_dataset(
    TRAIN_DATASET,
    split="train",
    streaming=True,
).shuffle(seed=SEED, buffer_size=1_000)

batches = get_batch(10, dataset) 
batch = next(batches) 
print(batch[0])
```
