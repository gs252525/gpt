```python
import glob, os, sysconfig, random

stdlib = sysconfig.get_paths()["stdlib"]
paths = sorted(glob.glob(os.path.join(stdlib, "**", "*"), recursive=True))
print(paths)
```

```python
files = []
for p in paths:
  try:
    t = open(p, encoding="utf-8").read()
  except Exception:
    continue
  files.append(t)
print(f"files: {len(files)}, chars: {sum(map(len, files))/1e6:.1f}M")
```

```python
print(files[:2])
```

```python
random.seed(0)
random.shuffle(files)
n_val = int(len(files) * 0.1)
val_files, train_files = files[:n_val], files[n_val:]
train_text, val_text = "\n\n".join(train_files), "\n\n".join(val_files)
```

```python
chars = sorted(list(set(train_text+val_text)))
vocab_size = len(chars)
stoi = {ch:i for i,ch in enumerate(chars)}
itos = {i:ch for i,ch in enumerate(chars)}
encode = lambda s: [stoi[c] for c in s]
decode = lambda l: "".join([itos[i] for i in l])
print(chars)
print(encode('a'), encode('b'))
print(decode([38,100,100,101]))
```

```python
train_data = torch.tensor(encode(train_text), dtype=torch.long)
val_data = torch.tensor(encode(val_text), dtype=torch.long)
```

```python
print(f"{train_data.shape}, {val_data.shape}")
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

def get_batch(split, batch_size, block_size):
  data = train_data if split == "train" else val_data
  ix = torch.randint(len(data) - block_size, (batch_size,))
  x = torch.stack([data[i:i+block_size] for i in ix])
  y = torch.stack([data[i+1:i+block_size+1] for i in ix])
  return x.to(device), y.to(device)
```

```python
import math
import torch.nn as nn
import torch.nn.functional as F
from dataclasses import dataclass

@dataclass
class Config:
  vocab_size: int
  block_size: int = 256
  n_layer: int = 6
  n_head: int = 6
  n_embd: int = 384
  dropout: float = 0.1

def causal_attention(q, k, v):
  # q, k, v : (B, n_head, T, head_dim)  T : 처리하는 길이
  # (B, T) -> (B, T, n_embd) -> (B, T, n_head, head_dim) -> (B, n_head, T, head_dim)
  T = q.size(-2)
  att = (q @ k.transpose(-2, -1)) / math.sqrt(q.size(-1))
  # (B, n_head, T, T)
  # i,j 행 : T_i @ T_j -> j<=i 에만 어텐션을 적용
  mask = torch.tril(torch.ones(T,T, device=q.device, dtype=torch.bool))
  att = att.masked_fill(~mask, float("-inf"))
  return F.softmax(att, dim=-1) @ v

class CausalSelfAttention(nn.Module):
  def __init__(self, c):
    super().__init__()
    assert c.n_embd % c.n_head == 0
    self.n_head = c.n_head
    self.qkv = nn.Linear(c.n_embd, 3 * c.n_embd)
    self.proj = nn.Linear(c.n_embd, c.n_embd)
    self.drop = nn.Dropout(c.dropout)

  def forward(self, x):
    B, T, C = x.shape
    q, k, v = self.qkv(x).split(C, dim=2) # B, T, 3*C -> (B,T,C)
    q, k, v = (t.view(B, T, self.n_head, C // self.n_head).transpose(1,2) for t in (q,k,v))
    y = causal_attention(q, k, v) # B, n_head, T, n_embd
    y = y.transpose(1, 2).contiguous().view(B,T,C)
    return self.drop(self.proj(y))

class Block(nn.Module):
  def __init__(self, c):
    super().__init__()
    self.ln1 = nn.LayerNorm(c.n_embd)
    self.attn = CausalSelfAttention(c)
    self.ln2 = nn.LayerNorm(c.n_embd)
    self.mlp = nn.Sequential(
        nn.Linear(c.n_embd, 4 * c.n_embd),
        nn.GELU(),
        nn.Linear(4 * c.n_embd, c.n_embd),
        nn.Dropout(c.dropout)
    )

  def forward(self, x):
    # x -> 대충 B,T,C -> 마지막 차원을 대략 분산 1로 고정시키기
    x = x + self.attn(self.ln1(x))
    # B, T, C -> B,T,C
    x = x + self.mlp(self.ln2(x))
    return x

class GPT(nn.Module):
  def __init__(self, c):
    super().__init__()
    self.c = c
    self.tok_emb = nn.Embedding(c.vocab_size, c.n_embd)
    self.pos_emb = nn.Embedding(c.block_size, c.n_embd)
    self.drop = nn.Dropout(c.dropout)
    self.blocks = nn.ModuleList(Block(c) for _ in range(c.n_layer))
    self.ln_f = nn.LayerNorm(c.n_embd)
    self.head = nn.Linear(c.n_embd, c.vocab_size)

    self.apply(self._init)

  def _init(self, m):
    if isinstance(m, nn.Linear):
        nn.init.normal_(m.weight, std=0.02)
        if m.bias is not None:
            nn.init.zeros_(m.bias)
    elif isinstance(m, nn.Embedding):
        nn.init.normal_(m.weight, std=0.02)

  def forward(self, idx, targets=None):
    B, T = idx.shape
    assert T <= self.c.block_size
    pos = torch.arange(T, device=idx.device)
    tok_emb = self.tok_emb(idx)
    pos_emb = self.pos_emb(pos)
    x = self.drop(tok_emb + pos_emb)
    for blk in self.blocks:
      x = blk(x)
    logits = self.head(self.ln_f(x))
    # B, T, n_embd ->
    # B, T
    # cross_entropy (개수, 정답) (개수, 로짓)
    loss = None
    if targets is not None:
      loss = F.cross_entropy(logits.view(-1, logits.size(-1)), targets.view(-1))
    return logits, loss
```

```python
# (a) attention 수치 검증
q, k, v = (torch.randn(2, 6, 16, 64) for _ in range(3))
ref = F.scaled_dot_product_attention(q, k, v, is_causal=True)
print("attention match:", torch.allclose(causal_attention(q, k, v), ref, atol=1e-5))

# (b) 모델 shape, 파라미터 수, 초기 loss
cfg = Config(vocab_size=vocab_size)
model = GPT(cfg).to(device)

# 1. 먼저 더미 데이터로 forward를 수행하여 LazyLinear의 파라미터를 실제로 생성(materialize)합니다.
with torch.no_grad():
    model(torch.zeros(1, cfg.block_size, dtype=torch.long, device=device))

# 2. 파라미터가 생성된 후 가중치 초기화(weight initialization)를 적용합니다.
model.apply(model._init)

print(f"params: {sum(p.numel() for p in model.parameters())/1e6:.1f}M")
x, y = get_batch("train", 8, cfg.block_size)
logits, loss = model(x, y)
print(logits.shape, f"loss: {loss.item():.3f} (expected ~ {math.log(vocab_size):.3f})")
```

```python
cfg0 = Config(vocab_size=vocab_size, dropout=0.0)
m0 = GPT(cfg0).to(device)
opt0 = torch.optim.AdamW(m0.parameters(), lr=1e-3)

x, y = get_batch("train", 8, cfg0.block_size)   # 고정 배치 하나
for i in range(601):
    _, loss = m0(x, y)
    opt0.zero_grad(set_to_none=True)
    loss.backward()
    opt0.step()
    if i % 40 == 0:
        print(i, f"{loss.item():.4f}")
del m0, opt0
```

```python
import time
from google.colab import drive
drive.mount("/content/drive")
CKPT = "/content/drive/MyDrive/code_gpt_ckpt.pt"

# 하이퍼파라미터
batch_size  = 64
max_iters   = 4000
eval_every  = 250
eval_iters  = 50
lr, min_lr  = 6e-4, 6e-5
warmup      = 100
grad_clip   = 1.0

# 모델 생성 -> materialize -> 초기화 -> 그 다음에 옵티마이저
cfg   = Config(vocab_size=vocab_size)
model = GPT(cfg).to(device)
with torch.no_grad():
    model(torch.zeros(1, cfg.block_size, dtype=torch.long, device=device))
model.apply(model._init)
print(f"params: {sum(p.numel() for p in model.parameters())/1e6:.1f}M")


decay    = [p for p in model.parameters() if p.dim() >= 2]   # 행렬/임베딩만 weight decay
no_decay = [p for p in model.parameters() if p.dim() < 2]    # bias, LayerNorm
opt = torch.optim.AdamW(
    [{"params": decay, "weight_decay": 0.1}, {"params": no_decay, "weight_decay": 0.0}],
    lr=lr, betas=(0.9, 0.95),
)
scaler = torch.amp.GradScaler("cuda")

def get_lr(it):
    if it < warmup:
        return lr * (it + 1) / warmup
    ratio = (it - warmup) / max(1, max_iters - warmup)
    return min_lr + 0.5 * (1 + math.cos(math.pi * ratio)) * (lr - min_lr)

@torch.no_grad()
def estimate_loss():
    model.eval()
    out = {}
    for split in ("train", "val"):
        ls = torch.zeros(eval_iters)
        for k in range(eval_iters):
            xb, yb = get_batch(split, batch_size, cfg.block_size)
            with torch.autocast("cuda", dtype=torch.float16):
                _, l = model(xb, yb)
            ls[k] = l.item()
        out[split] = ls.mean().item()
    model.train()
    return out

best_val, t0 = float("inf"), time.time()
model.train()
for it in range(max_iters + 1):
    for g in opt.param_groups:
        g["lr"] = get_lr(it)

    if it % eval_every == 0:
        L = estimate_loss()
        print(f"it {it:5d} | train {L['train']:.3f} | val {L['val']:.3f} | {time.time()-t0:.0f}s")
        if L["val"] < best_val:
            best_val = L["val"]
            torch.save({"model": model.state_dict(), "cfg": cfg.__dict__, "it": it,
                        "chars": chars, "val": best_val}, CKPT)

    xb, yb = get_batch("train", batch_size, cfg.block_size)
    with torch.autocast("cuda", dtype=torch.float16):
        _, loss = model(xb, yb)
    opt.zero_grad(set_to_none=True)
    scaler.scale(loss).backward()
    scaler.unscale_(opt)
    torch.nn.utils.clip_grad_norm_(model.parameters(), grad_clip)
    scaler.step(opt)
    scaler.update()
```

```python
@torch.no_grad()
def generate(prompt, max_new=400, temperature=0.8, top_k=40):
    model.eval()
    idx = torch.tensor([encode(prompt)], dtype=torch.long, device=device)
    for _ in range(max_new):
        ctx = idx[:, -cfg.block_size:]
        logits, _ = model(ctx)
        logits = logits[:, -1, :] / temperature
        if top_k:
            v, _ = torch.topk(logits, top_k)
            logits[logits < v[:, [-1]]] = float("-inf")
        probs = F.softmax(logits, dim=-1)
        idx = torch.cat([idx, torch.multinomial(probs, 1)], dim=1)
    model.train()
    return decode(idx[0].tolist())

print(generate("def "))
print("-" * 60)
print(generate("import os\n\nclass "))
```
